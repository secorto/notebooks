## Data types

Python is a dynamically typed programming language with strong typing. This means that we can create variables with any type of data without having to declare the type explicitly.

> However, Python has a mechanism ([Type hints](https://python.org)) 
> with which we can define the types to be used, 
> similar to what is done in TypeScript.

As you may have noticed in the previous variable declaration (`nombre: str = "Sergio"`), 
we added the annotation `: str`. 
In traditional versions of Python, typing was purely dynamic, 
which meant you could create variables without saying what type of data they were, 
even allowing them to change from text to number freely.

Starting with **Python 3.5+ (PEP 484)**, the language introduced **Type Hints**. 
This modern syntax lets you explicitly document what type of data a variable contains 
using a colon (`:`) followed by the data type (such as `str` for text strings, 
or `int` for integers).

*Learning note:* Python remains a flexible language 
(it will not throw an error if you store a number where you said a text should go), 
but including Type Hints is an excellent practice. 
It helps your code editor understand what you are doing better, 
offers smart autocompletion, and warns you if you are making a logic error 
before running the program.

## Import

There are times when the built-in data types in Python are not enough, so we need to import external "objects" such as functions or classes. In Python, code is organized into packages and modules.

Do you remember that example where we tried to add `0.1 + 0.2`? Well, to solve it, we are going to import the `Decimal` class from the `decimal` module, which will allow us to operate on numbers in base 10.

```python
from decimal import Decimal
Decimal('0.1') + Decimal('0.2')
```

In Python we can import the entire module.

```python
import decimal
decimal.Decimal('0.1') + decimal.Decimal('0.2')
```

We can even set aliases using the `as` keyword.

```python
import decimal as dec
dec.Decimal('0.1') + dec.Decimal('0.2')
```

## Lists

In Python, we can have collections of objects where we have a list of elements and need to add or remove values (i.e., mutable structures). For this, we can use lists (`list`), represented with square brackets.

```python
lista: list[str] = ["a", "b", "c", "a"]
print(lista)
```

# Create Functions

We can create functions using the `def` keyword followed by the function name we want to create. For example, we will create a `sum` function.

```python
def suma(a: int, b: int) -> int:
    return a + b

print(suma(3, 4))
print(suma(-1, 4))
```

# Conditionals

In Python, we can use conditionals to execute or skip a code block.

```python
def mayor_de_edad(edad: int) -> None:
    if(edad >= 18):
        print("nos tomamos unas polas?")
    elif(edad>=2):
        print("quieres un helado?")
    else:
        print("quien es un bebe bonito?")

mayor_de_edad(21)
mayor_de_edad(12)
mayor_de_edad(1)
```

# Loops

In Python, we can use repetitive loops to iterate over a collection. For example, we can use a **for** loop to calculate the squares of a list.

```python
numeros: list[int] = [2, 3, 6, 8]
cuadrados: list[int] = []

for numero in numeros:
    cuadrados.append(numero**2)

print(cuadrados)
```

Of course, there is also a condensed way to do the previous exercise using list comprehensions.

```python
[numero ** 2 for numero in numeros]
```

You can also create a repetitive **while** loop. For example, we can keep asking as long as we receive a value different from 's'.

```python
respuesta: str = 's'
while respuesta == 's':
    respuesta = input("Desea continuar? ")
```

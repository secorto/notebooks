## Tipos de datos

Python es un lenguaje de programación fuertemente tipado de tipado dinámico, 
esto quiere decir que podemos crear variables con cualquier tipo de datos 
sin preocuparnos por declarar explícitamente el tipo.

> Sin embargo, Python posee un mecanismo ([Type hints](https://python.org)) 
> con el que se puede definir los tipos que se van a usar, 
> algo así como lo que se hace con TypeScript.

Como habrás notado en la declaración de la variable anterior (`nombre: str = "Sergio"`), 
hemos añadido la anotación `: str`. 
En las versiones tradicionales de Python, el tipado era puramente dinámico, 
lo que significaba que podías crear variables sin decir qué tipo de dato eran, 
permitiendo incluso que cambiaran de texto a número de forma libre.

A partir de **Python 3.5+ (PEP 484)**, el lenguaje introdujo los **Type Hints**. 
Esta sintaxis moderna te permite documentar de forma explícita qué tipo de datos contiene una variable 
usando dos puntos (`:`) seguido del tipo de dato (como `str` para cadenas de texto, 
o `int` para números enteros).

*Nota de aprendizaje:* Python sigue siendo un lenguaje flexible 
(no arrojará un error si guardas un número donde dijiste que iba un texto), 
pero incluir Type Hints es una excelente práctica. 
Ayuda a que tu editor de código entienda mejor qué estás haciendo, 
te ofrezca un autocompletado inteligente y te avise si estás cometiendo un error de lógica 
antes de ejecutar el programa.

## Import

Hay veces que los tipos de datos que vienen pre cargados en python no nos son suficientes para lo cual necesitamos importar "objetos" externos por ejemplo funciones o bien clases, en python el código esta organizado en paquetes y módulos


Recuerdan ese ejemplo donde intentabamos sumar `0.1 + 0.2`, bueno para poderlo resolver vamos a importar la clase `Decimal` del módulo `decimal` el cual nos va a permitir operar sobre números en base 10

```python
from decimal import Decimal
Decimal('0.1') + Decimal('0.2')
```

En python podemos importar el módulo completo

```python
import decimal
decimal.Decimal('0.1') + decimal.Decimal('0.2')
```

Incluso podemos poner alias con la palabra `as`

```python
import decimal as dec
dec.Decimal('0.1') + dec.Decimal('0.2')
```

# Crear Funciones

Podemos crear funciones usando la palabra reservada def con el nombre de la función que deseamos crear, por ejemplo crearemos una función suma

```python
def suma(a: int, b: int) -> int:
    return a + b

print(suma(3, 4))
print(suma(-1, 4))
```

# Condicionales

En python podemos hacer uso de condicionales para ejecutar o no un fragmento de código

```python
def mayor_de_edad(edad: int) -> None:
    if(edad >= 18):
        print("nos tomamos unas polas?")
    elif(edad>=2):
        print("quieres un helado?")
    else:
        print("quien es un bebe bonito?")

mayor_de_edad(21)
mayor_de_edad(12)
mayor_de_edad(1)
```

# Ciclos

En python podemos usar ciclos repetitivos para iterar una colección, por ejemplo podemos usar un **for** para sacar los cuadrados de una lista

```python
numeros: list[int] = [2, 3, 6, 8]
cuadrados: list[int] = []

for numero in numeros:
    cuadrados.append(numero**2)

print(cuadrados)
```

Claro, que tambien hay una forma condensada de hacer el ejercicio anterior que puede ser usando list comprehension

```python
[numero ** 2 for numero in numeros]
```

Tambien se puede crear un ciclo repetitivo **while**, por ejemplo podemos seguir preguntando siempre si nos dan un valor distinto de 's'

```python
respuesta: str = 's'
while respuesta == 's':
    respuesta = input("Desea continuar? ")
```
